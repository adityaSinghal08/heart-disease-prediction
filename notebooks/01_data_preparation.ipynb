```python
# import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the train dataset
df = pd.read_csv('../data/train.csv')
```

```python
# check data head
df.head()
```

```python
# check dataset info
df.info()
```

```python
# drop the 'id' column as it is not useful for prediction
df.drop('id', axis=1, inplace=True)
```

```python
# label encode the target (Heart Disease) variable
df['Heart Disease'] = df['Heart Disease'].map({'Presence': 1, 'Absence': 0})
```

```python
# delete duplicate rows if any
original_shape = df.shape
print(f"Original shape: {original_shape}")

df.drop_duplicates(inplace=True)

new_shape = df.shape
if original_shape == new_shape:
    print("No duplicate rows found.")
else:
    print(f"{original_shape[0] - new_shape[0]} duplicate rows removed.")
    print(f"New shape: {new_shape}")
```

```python
# save the cleaned dataset as cleaned_train.csv
df.to_csv('../data/cleaned_train.csv', index=False)
```
