```python
# import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
```

```python
# load the feature engineered training dataset
df = pd.read_csv("../data/fe_train.csv")
```

```python
# check data info
df.info()
```

```python
# check data description
df.describe().T
```

```python
# check count of unique values in each column
df.nunique().sort_values()
```

```python
# print all unique values for columns with less than 5 unique values
for col in df.columns:
    if df[col].nunique() < 5:
        print(f"{col}: {df[col].unique()}")
```

```python
# define numeric (continuous/discrete), categorical (binary/nominal/ordinal), and target columns

TARGET = ["Heart Disease"]

CATEGORICAL_BINARY = [
    "Sex",
    "FBS over 120",
    "Exercise angina",
    "Any_vessel_disease",
    "Severe_vessel_disease",
    "Abnormal_thallium",
    "Exercise_ischemia_flag",
    "Abnormal_EKG_flag",
    "Typical_angina_flag",
    "Low_HR_response_flag",
    "High_BP_flag",
    "High_Chol_flag"
]

CATEGORICAL_NOMINAL = [
    "Chest pain type",
    "EKG results",
    "Thallium"
]

CATEGORICAL_ORDINAL = [
    "Slope of ST",
    "ST_severity_level",
    "Chest_pain_risk_score",
    "Age_group"
]

NUMERIC_CONTINUOUS = [
    "ST depression",
    "ST_depression_normalized",
    "BP_Chol_ratio"
]

NUMERIC_DISCRETE = [
    "Age",
    "BP",
    "Cholesterol",
    "Max HR",
    "Heart_rate_reserve",
    "Age_squared",
    "Age_Sex_interaction",
    "Number of vessels fluro",
    "Metabolic_risk_score",
    "Exercise_risk_score"
]
```

```python
# create a grid with countplots for categorical variables with hue as target variable
plt.figure(figsize=(20, 20))
for i, col in tqdm(enumerate(CATEGORICAL_BINARY + CATEGORICAL_NOMINAL + CATEGORICAL_ORDINAL)):
    plt.subplot(7, 3, i + 1)
    sns.countplot(data=df, x=col, hue="Heart Disease")
    plt.title(f"Countplot of {col} with Heart Disease")
    plt.tight_layout()
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.legend(title="Heart Disease", loc="upper right")
```

```python
# calculate the chance of heart disease for each category in the categorical variables
for col in CATEGORICAL_BINARY + CATEGORICAL_NOMINAL + CATEGORICAL_ORDINAL:
    print(f"Chance of heart disease for each category in {col}:")
    print(df.groupby(col)["Heart Disease"].mean())
    print("\n")
```

```python
# create a grid with kde distribution plots for continuous numeric variables with hue as target variable
plt.figure(figsize=(14, 14))
for i, col in tqdm(enumerate(NUMERIC_CONTINUOUS)):
    plt.subplot(2, 2, i + 1)
    sns.kdeplot(data=df, x=col, hue="Heart Disease", fill=True)
    plt.title(f"KDE plot of {col} with Heart Disease")
    plt.tight_layout()
    plt.xlabel(col)
    plt.ylabel("Heart Disease")
    plt.legend(title="Heart Disease", loc="upper right")
```

```python
# create a grid with violin plots for target variables against discrete numeric variables
plt.figure(figsize=(20, 20))
for i, col in tqdm(enumerate(NUMERIC_DISCRETE)):
    plt.subplot(4, 3, i + 1)
    sns.violinplot(data=df, x="Heart Disease", y=col)
    plt.title(f"Violin plot of {col} with Heart Disease")
    plt.tight_layout()
    plt.xlabel("Heart Disease")
    plt.ylabel(col)
```

```python
# create correlation heatmap for numeric variables and target variable
plt.figure(figsize=(12, 10))
numeric_cols = NUMERIC_CONTINUOUS + NUMERIC_DISCRETE + TARGET
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Numeric Variables and Target Variable")
```
